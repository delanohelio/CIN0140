## 1) Definição de grafo

Um grafo $G = (V, E)$ tem:
- $V$: conjunto de vértices (ou nós).
- $E$: conjunto de arestas ligando pares de vértices.

Em grafos não direcionados, `(u, v)` representa a mesma aresta que `(v, u)`.
Em dígrafos, a direção importa: `u -> v` é diferente de `v -> u`.

```python
# Exemplo simples: grafo não direcionado por lista de adjacência
grafo = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D', 'E'],
    'D': ['B', 'C', 'E'],
    'E': ['C', 'D']
}

for v, vizinhos in grafo.items():
    print(f"{v}: {vizinhos}")
```

## 2) Representações: matriz e lista de adjacência

Segundo o capítulo 1.4, duas representações são muito usadas:
- **Matriz de adjacência**: matriz $n \times n$ com 0/1 (ou pesos).
- **Lista de adjacência**: para cada vértice, uma lista de vizinhos.

Ideia prática:
- Matriz: mais simples para testar aresta em $O(1)$, mas gasta mais memória.
- Lista: mais econômica em grafos esparsos.

```python
def lista_para_matriz(g):
    vertices = sorted(g.keys())
    indice = {v: i for i, v in enumerate(vertices)}
    n = len(vertices)
    A = [[0] * n for _ in range(n)]

    for u in vertices:
        for v in g[u]:
            A[indice[u]][indice[v]] = 1

    return vertices, A


vertices, A = lista_para_matriz(grafo)
print("Vértices:", vertices)
print("Matriz de adjacência:")
for linha in A:
    print(linha)
```

## 3) Grafos ponderados

Em grafos ponderados, cada aresta tem um peso (custo).

Exemplos de uso:
- menor caminho em mapas
- roteamento em redes
- problemas de otimização

```python
# Lista de adjacência com pesos: (vizinho, peso)
grafo_pond = {
    'A': [('B', 5), ('C', 2)],
    'B': [('A', 5), ('D', 4)],
    'C': [('A', 2), ('D', 1), ('E', 7)],
    'D': [('B', 4), ('C', 1), ('E', 3)],
    'E': [('C', 7), ('D', 3)]
}

for u, adj in grafo_pond.items():
    print(f"{u}: {adj}")
```

## 4) Caminho, ciclo e conectividade

Conceitos importantes do capítulo 1.4:
- **Caminho**: sequência de vértices adjacentes.
- **Ciclo**: caminho com comprimento positivo que começa e termina no mesmo vértice.
- **Grafo conexo**: existe caminho entre todo par de vértices.

```python
def bfs_visitados(g, inicio):
    fila = deque([inicio])
    visitados = {inicio}

    while fila:
        u = fila.popleft()
        for v in g[u]:
            if v not in visitados:
                visitados.add(v)
                fila.append(v)

    return visitados


def eh_conexo(g):
    if not g:
        return True
    inicio = next(iter(g))
    return len(bfs_visitados(g, inicio)) == len(g)


def tem_ciclo_undirecionado(g):
    visitados = set()

    def dfs(u, pai):
        visitados.add(u)
        for v in g[u]:
            if v not in visitados:
                if dfs(v, u):
                    return True
            elif v != pai:
                return True
        return False

    for u in g:
        if u not in visitados and dfs(u, None):
            return True

    return False


print("Grafo é conexo?", eh_conexo(grafo))
print("Grafo tem ciclo?", tem_ciclo_undirecionado(grafo))
```

## 5) Percursos em grafos: BFS e DFS

Como visto em aula, os dois percursos clássicos são:
- **BFS** (busca em largura): explora por níveis.
- **DFS** (busca em profundidade): aprofunda antes de voltar.

```python
def bfs_ordem(g, inicio):
    fila = deque([inicio])
    visitados = {inicio}
    ordem = []

    while fila:
        u = fila.popleft()
        ordem.append(u)
        for v in g[u]:
            if v not in visitados:
                visitados.add(v)
                fila.append(v)

    return ordem


def dfs_ordem(g, inicio):
    ordem = []
    visitados = set()

    def dfs(u):
        visitados.add(u)
        ordem.append(u)
        for v in g[u]:
            if v not in visitados:
                dfs(v)

    dfs(inicio)
    return ordem


print("Ordem BFS a partir de A:", bfs_ordem(grafo, 'A'))
print("Ordem DFS a partir de A:", dfs_ordem(grafo, 'A'))
```
