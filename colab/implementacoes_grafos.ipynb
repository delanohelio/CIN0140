# Implementacoes de Grafos (Matriz de Adjacencia)

Neste notebook, implementamos uma classe de grafo baseada em **matriz de adjacencia**
com as operacoes pedidas no enunciado.

## Operacoes implementadas

- `int n(G g)`
- `int e(G g)`
- `int first(G g, int v)`
- `int next(G g, int v, int w)`
- `void setEdge(G g, int i, int j, int wt)`
- `void delEdge(G g, int i, int j)`
- `boolean isEdge(G g, int i, int j)`
- `int weight(G g, int i, int j)`
- `void setMark(G g, int v, int val)`
- `int getMark(G g, int v)`

Observacao: indices de vertices vao de `0` ate `n-1`.

```python
class GrafoMatriz:
    def __init__(self, num_vertices, directed=True):
        if num_vertices <= 0:
            raise ValueError("O numero de vertices deve ser positivo.")

        self._n = num_vertices
        self._directed = directed
        self._mat = [[0] * num_vertices for _ in range(num_vertices)]
        self._marks = [0] * num_vertices
        self._e = 0

    def _validar_vertice(self, v):
        if not (0 <= v < self._n):
            raise IndexError(f"Vertice invalido: {v}.")

    # int n(G g): numero de vertices
    def n(self):
        return self._n

    # int e(G g): numero de arestas
    def e(self):
        return self._e

    # int first(G g, int v): primeiro vizinho de v
    def first(self, v):
        self._validar_vertice(v)
        for w in range(self._n):
            if self._mat[v][w] != 0:
                return w
        return -1

    # int next(G g, int v, int w): proximo vizinho de v apos w
    def next(self, v, w):
        self._validar_vertice(v)
        if w < -1 or w >= self._n:
            raise IndexError(f"Vertice de referencia invalido: {w}.")

        for x in range(w + 1, self._n):
            if self._mat[v][x] != 0:
                return x
        return -1

    # void setEdge(G g, int i, int j, int wt): adicionar aresta
    def setEdge(self, i, j, wt):
        self._validar_vertice(i)
        self._validar_vertice(j)

        if wt == 0:
            raise ValueError("Peso 0 e reservado para ausencia de aresta.")

        if not self.isEdge(i, j):
            self._e += 1

        self._mat[i][j] = wt

        if not self._directed and i != j:
            self._mat[j][i] = wt

    # void delEdge(G g, int i, int j): remover aresta
    def delEdge(self, i, j):
        self._validar_vertice(i)
        self._validar_vertice(j)

        if self.isEdge(i, j):
            self._e -= 1

        self._mat[i][j] = 0

        if not self._directed and i != j:
            self._mat[j][i] = 0

    # boolean isEdge(G g, int i, int j): testa se existe aresta
    def isEdge(self, i, j):
        self._validar_vertice(i)
        self._validar_vertice(j)
        return self._mat[i][j] != 0

    # int weight(G g, int i, int j): retorna peso
    def weight(self, i, j):
        self._validar_vertice(i)
        self._validar_vertice(j)
        return self._mat[i][j]

    # void setMark(G g, int v, int val): usado para DFS/BFS
    def setMark(self, v, val):
        self._validar_vertice(v)
        self._marks[v] = val

    # int getMark(G g, int v): usado para DFS/BFS
    def getMark(self, v):
        self._validar_vertice(v)
        return self._marks[v]

    def __str__(self):
        linhas = ["Matriz de adjacencia:"]
        for linha in self._mat:
            linhas.append(str(linha))
        return "\n".join(linhas)
```

```python
# Funcoes auxiliares com os nomes do enunciado
def n(g):
    return g.n()


def e(g):
    return g.e()


def first(g, v):
    return g.first(v)


def next(g, v, w):
    return g.next(v, w)


def setEdge(g, i, j, wt):
    g.setEdge(i, j, wt)


def delEdge(g, i, j):
    g.delEdge(i, j)


def isEdge(g, i, j):
    return g.isEdge(i, j)


def weight(g, i, j):
    return g.weight(i, j)


def setMark(g, v, val):
    g.setMark(v, val)


def getMark(g, v):
    return g.getMark(v)
```

```python
# Exemplo de uso (digrafo)
g = GrafoMatriz(5, directed=True)

setEdge(g, 0, 1, 10)
setEdge(g, 0, 3, 7)
setEdge(g, 1, 4, 2)
setEdge(g, 3, 2, 5)

print("n(g):", n(g))
print("e(g):", e(g))
print("first(g, 0):", first(g, 0))
print("next(g, 0, 1):", next(g, 0, 1))
print("isEdge(g, 0, 3):", isEdge(g, 0, 3))
print("weight(g, 0, 3):", weight(g, 0, 3))

setMark(g, 2, 1)
print("getMark(g, 2):", getMark(g, 2))

delEdge(g, 0, 1)
print("e(g) apos delEdge(0,1):", e(g))
print(g)
```

## Observacoes finais

- Se quiser representar grafo **nao direcionado**, use `directed=False` ao criar o grafo.
- Nesse caso, `setEdge(i, j, wt)` atualiza tambem `j -> i` automaticamente.
- `first` e `next` retornam `-1` quando nao ha mais vizinhos.
